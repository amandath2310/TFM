```python
import os
import sys
from sys import argv
import numpy as np
import chemcoord as cc
from numpy.core.numeric import full

#Displacement variables
# This variables represent the increasing of displacement (en las distancias y angulos de torsion)
disp = 0.05
disp_angle = 10.0
disp_angle2 = 5.0
# Save the standar output and the will be restaured later
orig_stdout = sys.stdout
```

```python
ethylene = cc.Cartesian.read_xyz('./geom.xyz', start_index=1)
#Reading in FC structure and generating Z-matrix file

zethylene = ethylene.get_zmat()

# Creation of a new zmatrix

print(zethylene)
```

```python
geoms = []
# Creating a coppy of the z-matrix
zethylene2 = zethylene.copy()

#Selecting the non-redundant internal coords to modify
bond_before_assignment = zethylene.loc[1,'bond']
dihedral1_before_assignment = zethylene.loc[3,'dihedral']
dihedral2_before_assignment = zethylene.loc[4,'dihedral']
angle1_before_assignment = zethylene.loc[3,'angle']
angle2_before_assignment = zethylene.loc[4,'angle']
angle1_before_assignment = 132.5
angle2_before_assignment = 132.5

#Looping over the different distortions
for i in range(-2,10): # Going from -0.1 Angs to + 0.6 Angst stretches
   zethylene2.safe_loc[1,'bond'] = bond_before_assignment + i * disp
   for j in range(-9,0): # Ranging from 0 to 90 degrees
      zethylene2.safe_loc[3,'dihedral'] = dihedral1_before_assignment + j * disp_angle2
      zethylene2.safe_loc[3,'angle'] = angle1_before_assignment + j * disp_angle2
      zethylene2.safe_loc[4,'dihedral'] = dihedral2_before_assignment - j * disp_angle2
      zethylene2.safe_loc[4,'angle'] = angle2_before_assignment + j * disp_angle2
      xyz1 = zethylene2.get_cartesian().to_xyz()
      geoms.append(xyz1)

    #printing out to individual xyz-type files
      sys.stdout = open("./dist.stretch."+str('{:.3f}'.format(i*disp+bond_before_assignment))+".torsion."+str(j*disp_angle2), "w")
      print(xyz1)
      sys.stdout.close()

```
